# Logistic Regression

```python
import numpy as np

from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
X, y = data.data, data.target
```

```python
def normalize_features(X):
    """
    Standardize each column of X:
      X_norm = (X - mean) / std
    Returns: (X_norm, mu, sigma)
    """
    mu    = np.mean(X, axis=0)
    sigma = np.std(X,  axis=0)
    X_norm = (X - mu) / sigma
    return X_norm, mu, sigma
    
```

```python
def compute_cost(X, y, theta):
    """MSE cost for linear regression."""
    m    = len(y)
    pred = X.dot(theta)
    return (1/(2*m)) * np.sum((pred - y)**2)

def gradient_descent(X, y, theta, alpha, num_iters):
    """Batch gradient descent for linear regression."""
    m = len(y)
    J_history = []
    for _ in range(num_iters):
        error    = X.dot(theta) - y
        gradient = (1/m)*X.T.dot(error)
        theta    = theta - alpha*gradient
        J_history.append(compute_cost(X, y, theta))
    return theta, J_history
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_cost_logistic(X, y, theta, lambda_=0.0):
    """Regularized logistic cost (binary cross-entropy)."""
    m = len(y)
    h = sigmoid(X.dot(theta))
    ce   = -(1/m)*(y.dot(np.log(h)) + (1-y).dot(np.log(1-h)))
    reg  = (lambda_/(2*m))*np.sum(theta[1:]**2)  # exclude bias
    return ce + reg

def gradient_descent_logistic(X, y, theta, alpha, num_iters, lambda_=0.0):
    """Batch gradient descent for logistic regression w/ L2."""
    m = len(y)
    J_history = []
    for _ in range(num_iters):
        h     = sigmoid(X.dot(theta))
        error = h - y
        grad  = (1/m)*X.T.dot(error)
        grad[1:] += (lambda_/m)*theta[1:]
        theta = theta - alpha*grad
        J_history.append(compute_cost_logistic(X, y, theta, lambda_))
    return theta, J_history
```

```python
# Generate synthetic classification data
np.random.seed(1)
Xc = np.random.randn(100, 2)
yc = (Xc[:,0] + Xc[:,1] > 0).astype(int)

# Feature scaling
Xc_scaled, mu_c, sigma_c = normalize_features(Xc)
X_log = np.c_[ np.ones((100,1)), Xc_scaled ]

# Initialize parameters
theta_init_log = np.zeros(3)

# Run gradient descent
theta_log, J_hist_log = gradient_descent_logistic(
    X_log, yc, theta_init_log,
    alpha=0.1, num_iters=500,
    lambda_=1.0
)

print("Logistic Regression learned parameters:", theta_log)
print("Final cost (cross-entropy):", J_hist_log[-1])
```
